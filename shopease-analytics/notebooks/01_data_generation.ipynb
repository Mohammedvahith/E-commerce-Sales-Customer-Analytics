# Products.csv

```python
num_products  = 500

categories = {
    "Electronics": ["Headphones", "Smartphones", "Laptops", "Accessories"],
    "Fashion": ["Shoes", "Clothing", "Bags", "Watches"],
    "Home": ["Furniture", "Kitchen", "Decor", "Lighting"],
    "Beauty": ["Skincare", "Makeup", "Haircare", "Fragrance"],
    "Sports": ["Fitness", "Outdoor", "Sportswear", "Equipment"]
}
```

```python
product_rows = []

for i in range(1,num_products + 1):

    category = np.random.choice(
        list(categories.keys())
    )

    sub_category = np.random.choice(
        categories[category]
    )

    product_rows.append({
        "product_id" : f"P{i:05d}",
        "product_name" : f"{sub_category} Product {i}",
        "category" : category,
        "sub_category" : sub_category
    })

products =pd.DataFrame(product_rows)

products.head()
```

```python
products["price"] = np.round(
    np.random.uniform(20,2000,num_products),
    2
)
```

```python
products.head()
```

```python
products["cost"] = np.round(
    products["price"] *np.random.uniform(
        0.50,0.80,num_products
    ),
    2
)
```

```python
products.head()
```

```python
products["profit_per_unit"] = (
    products["price"] - products["cost"]
).round(2)
```

```python
products.head()
```

```python
num_payments = len(orders)
```

```python
payment_methods = np.random.choice([
    "Credit Card",
    "Debit Card",
    "Cash on Delivery",
    "Digital Wallet",
    "Bank Transfer"
],
size = num_payments,
p = [0.30,0.25,0.20,0.20,0.05]
)
```

```python
payment_status_map = {
    "Delivered": "Completed",
    "Cancelled": "Cancelled",
    "Returned": "Refunded",
    "Pending": "Pending"
}
```

```python
payment_statuses = orders["order_status"].map(payment_status_map)
```

```python
order_totals = (
    order_items
    .groupby("order_id")["total_amount"]
    .sum()
    .round(2)
)
```

```python
print(order_totals)
```

```python
payment_amounts = orders["order_id"].map(
    order_totals
)
```
