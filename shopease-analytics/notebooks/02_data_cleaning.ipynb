# Checking missing values in datasets

```python
datasets = {
    "Customers" : customers,
    "Products" : products,
    "Orders" : orders,
    "Order_items" : order_items,
    "Payments" : payments
}

for name, df in datasets.items():
    print(f"\n{name}")
    print(df.isna().sum())
```

# Count and Calculate the percentage of missing values 

```python
for name, df in datasets.items():
    missing_count = df.isna().sum().sum()
    total_cell = df.size
    missing_percentage = (missing_count/total_cell) * 100

    print(f"{name} : ")
    print(f"Missing cells : {missing_count}")
    print(f"Missing percentage : {missing_percentage:.2f}%")
    print("-" * 30)
```

# Verify the financial calculations

```python
# Check for Gross amount
expected_gross = (order_items["quantity"]*order_items["unit_price"]).round(2)

gross_mismatch = (
    expected_gross - order_items["gross_amount"]
).abs() > 0.01


print(
    "Base Exception Gross amount mismatches : ",gross_mismatch.sum()
)
```

```python
# check for discount amount 
expected_discount = ((order_items["discount"]/100) * order_items["gross_amount"]).round(2)

discount_mismatch = (
    expected_discount - order_items["discount_amount"]
).round(2).abs() > 0.01 # Corrected the error by adding the round(2)

print(
    "Base Exception discount amount mismatches : ",discount_mismatch.sum()
)
```

```python
# check for final amount 
expected_total = (
    expected_gross - expected_discount
).round(2)

total_mismatch = (
    expected_total - order_items["total_amount"]
).round(2).abs() > 0.01  # Corrected the error by adding the round(2)

print(
    "Total amount mismatches : ",total_mismatch.sum()
)
```

```python
# Mismatch Checking 
# STEP - 1
order_items["expected_total"] = expected_total

display(
    order_items.loc[
        total_mismatch,
    [
        "quantity",
        "unit_price",
        "gross_amount",
        "discount_amount",
        "total_amount",
        "expected_total"
    ]
].head(10)
)
```

```python
# Mismatch Checking 
# STEP - 2
# Compare stored discount with expected discount
discount_difference = (
    expected_discount - order_items["discount_amount"]
).round(2).abs()

print("Discount mismatches:", (discount_difference > 0.01).sum())

print("Maximum discount difference:", discount_difference.max())
```

```python
order_totals = (
    order_items.groupby("order_id")["total_amount"]
    .sum()
    .round(2)
    .reset_index(name="calculated_order_total")
)

order_totals.head()
```

```python
payment_amount_check = payments.merge(
    order_totals,
    on = "order_id",
    how = "left",
    validate = "one_to_one"
)

display(payment_amount_check.head())

payment_amount_check["amount_difference"] = (
    payment_amount_check["payment_amount"]
    -
    payment_amount_check["calculated_order_total"]
).round(2)

display(payment_amount_check.head())
```

```python
payment_amount_mismatch = (payment_amount_check["amount_difference"].abs() > 0.01)

display(payment_amount_mismatch.head())
```

```python
print("payment amount mismatches : ",
      payment_amount_mismatch.sum())
```

```python
display(
    payment_amount_check.loc[
        payment_amount_mismatch,
            [
                "order_id",
                "payment_amount",
                "calculated_order_total",
                "amount_difference"
            ]    
            ].head()
)
```

```python
print("Order Status : \n")
print(orders["order_status"].value_counts())

print("\nPayment Status : \n")
print(payments["payment_status"].value_counts())
```

```python
status_check = orders[
    [
        "order_id",
        "order_status"
    ]
].merge(
    payments[["order_id","payment_status"]],
    on = "order_id",
    how = "left",
    validate = "one_to_one"
)

status_check.head()
```

```python
expected_status = {
    "Delivered":"Completed",
    "Cancelled":"Cancelled",
    "Returned":"Refunded",
    "Pending":"Pending"
}
```

```python
# Merge
status_check["expected_payment_status"] = (
    status_check["order_status"].map(expected_status)
)

status_check.head()
```

```python
# Check for the Mismatch

status_mismatch = status_check[
    status_check["payment_status"] != status_check["expected_payment_status"]
]

print("Status mismatches : ",len(status_mismatch))

display(status_mismatch.head(10))
```
