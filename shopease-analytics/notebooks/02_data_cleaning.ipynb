# Checking missing values in datasets

```python
datasets = {
    "Customers" : customers,
    "Products" : products,
    "Orders" : orders,
    "Order_items" : order_items,
    "Payments" : payments
}

for name, df in datasets.items():
    print(f"\n{name}")
    print(df.isna().sum())
```

# Count and Calculate the percentage of missing values 

```python
for name, df in datasets.items():
    missing_count = df.isna().sum().sum()
    total_cell = df.size
    missing_percentage = (missing_count/total_cell) * 100

    print(f"{name} : ")
    print(f"Missing cells : {missing_count}")
    print(f"Missing percentage : {missing_percentage:.2f}%")
    print("-" * 30)
```

```python
id_checks = {
    "customers" : (customers,"customer_id"),
    "orders" : (orders,"order_id"),
    "order_items" : (order_items,"order_item_id"),
    "products" : (products,"product_id"),
    "payments" : (payments,"payment_id")
}

for name ,(df ,id_column) in id_checks.items():
    duplicate_count = df.duplicated(
        subset = [id_column],
        keep = False
    ).sum()
    print(f"{name} : {duplicate_count} rows with repeated IDs")
```

# Validate Date Relationships

```python
customer_dates = customers[["customer_id","signup_date"]]

customer_dates.head()
```

```python
orders.head()
```

```python
order_check = orders.merge(
    customer_dates,
    on="customer_id",
    how="left",
    validate="many_to_one"
)

order_check.head()
```

```python
# After Merging the dataset checking for any null or mismatch
print(
    "Orders without a matching customer : ",
    order_check["signup_date"].isna().sum()  )
```

```python
invalid_orders = order_check[
    order_check["orders_date"] < order_check["signup_date"]
]
print("Order placed before signup : ",len(invalid_orders))

display(invalid_orders.head())
```
